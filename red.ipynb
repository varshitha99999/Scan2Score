```python
import cv2
import numpy as np

img = cv2.imread("input/RedInkDetection.jpg")

# LAB conversion
lab = cv2.cvtColor(img, cv2.COLOR_BGR2LAB)
L, A, B = cv2.split(lab)

A_norm = cv2.normalize(A, None, 0, 255, cv2.NORM_MINMAX)

# Adaptive threshold (this part was GOOD)
mean_A = np.mean(A_norm)
std_A = np.std(A_norm)
thresh_val = mean_A + 1.2 * std_A

_, red_mask = cv2.threshold(
    A_norm, thresh_val, 255, cv2.THRESH_BINARY
)

# 🔑 NEW: median blur removes dots without killing strokes
red_mask = cv2.medianBlur(red_mask, 5)

# VERY light morphology
kernel = np.ones((3,3), np.uint8)
red_mask = cv2.morphologyEx(red_mask, cv2.MORPH_OPEN, kernel, 1)

# Extract red marks
red_only = cv2.bitwise_and(img, img, mask=red_mask)

cv2.imshow("Original", img)
cv2.imshow("Red Mask", red_mask)
cv2.imshow("Red Marks Only", red_only)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
